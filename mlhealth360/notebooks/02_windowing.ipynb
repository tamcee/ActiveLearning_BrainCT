# HU windowing
Verify the three windows on known HU values and a real CT slice.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.data.windowing import window_slice

hu = np.linspace(-1000, 1500, 512*512, dtype=np.float32).reshape(512,512)
channels = window_slice(hu)
fig, ax = plt.subplots(1,3,figsize=(12,4))
for i, name in enumerate(['Brain','Subdural','Bone']):
    ax[i].imshow(channels[i], cmap='gray', vmin=0, vmax=1); ax[i].set_title(name); ax[i].axis('off')
plt.show()
```
