# Baseline experiment
Run the configured training CLI from repository root, then inspect history and metrics.

```python
import json, pandas as pd, matplotlib.pyplot as plt
history_path='experiments/baseline/history.json'
with open(history_path) as f: history=json.load(f)
df=pd.DataFrame(history)
display(df.tail())
df.plot(x='epoch', y=['train_loss','val_loss']); plt.show()
df.plot(x='epoch', y='val_macro_dice'); plt.show()
```
