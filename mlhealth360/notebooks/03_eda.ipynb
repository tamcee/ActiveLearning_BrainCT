# Dataset EDA
Summarize per-study, per-slice and per-class imbalance before training.

```python
from pathlib import Path
import numpy as np, nibabel as nib, pandas as pd
from src.data.dataset import pair_nifti_files

image_map, mask_map = pair_nifti_files('data/train/images','data/train/masks')
rows=[]
for name, path in mask_map.items():
    m=np.asanyarray(nib.load(str(path)).dataobj)
    for z in range(m.shape[2]):
        vals, counts=np.unique(m[:,:,z], return_counts=True)
        d={int(k):int(v) for k,v in zip(vals,counts)}
        rows.append({'study':name,'slice':z, **{f'class_{c}_pixels':d.get(c,0) for c in range(6)}})
df=pd.DataFrame(rows)
display(df.head())
display(df[[f'class_{c}_pixels' for c in range(6)]].sum())
```
